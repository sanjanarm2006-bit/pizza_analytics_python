## data cleaning and preprocessing


```python
# check for null values
print(df.isnull().sum())
```

```python
# check for duplicates
print(df.duplicated().sum())
```

```python
df.dtypes
```

```python
# order_date and order_time --- convert it to date time format -- if error occurs then use format='mixed'
df['order_date'] = pd.to_datetime(df['order_date'],dayfirst=True,format='mixed') 
```

```python
# df['order_time'] = pd.to_datetime(df['order_time'],dayfirst=True,format='%H:%M:%S') 
df['order_time'] = pd.to_datetime(df['order_time']) 
```

```python
df.dtypes
```

```python
# Extracting month,date and year from order_date
# year
df['year'] = df['order_date'].dt.year
```

```python
# month -- month number -- dt.month
# month name -- dt.month_name()
df['month_number'] = df['order_date'].dt.month
```

```python
df['month_name'] = df['order_date'].dt.month_name()
```

```python
df.head()
```

```python
# day extraction
# day number-- dt.day
# day name-- dt.day_name()
df['day_number'] = df['order_date'].dt.day
```

```python
df['day_name'] = df['order_date'].dt.day_name()
```

```python
df.head()
```

```python
# Extracting hour, minutes and seconds
df['hour'] = df['order_time'].dt.hour
```

```python
# df['minute'] = df['order_time'].dt.minute
# df['seconds'] = df['order_time'].dt.second
```

```python
df['quarter'] = df['order_date'].dt.quarter
```

```python
df.columns
```

```python
# remove unnecessary columns-- year, month_number, day_number
df.drop(['year','month_number','day_number'],axis=1,inplace=True)
```

## groupby analysis

```python
# total quantity sold in every pizza category
a = df.groupby('pizza_category')['quantity'].sum()
print(a)
```

```python
# total quantity sold and total revenue in every pizza category
a = df.groupby('pizza_category')[['quantity','total_price']].sum()
print(a)
```

```python
# total orders in each quantity (based on order_id)
k = df.groupby('quantity')['order_id'].nunique()
print(k)
```

```python
k = df.groupby('quantity')['order_id'].count()
print(k)
```

```python
# calculate total revenue for each month
s = df.groupby('month_name')['total_price'].sum()
print(s)
```

```python
# calculate total orders for every day of week
r = df.groupby('day_name')['order_id'].nunique()
print(r)
```

```python
# calculate total quantity sold in every hour
qty = df.groupby('hour')['quantity'].sum()
m = qty.index
n = qty.values
plt.plot(m,n,'bD:')
plt.title("total quantity sold in every hour")
plt.xlabel("hour")
plt.ylabel("total quantity sold")
plt.show()
```

```python
# top 5 pizzas based on revenue
top = df.groupby('pizza_name')['total_price'].sum().sort_values().tail(5)
print(top)
```

```python
# top 5 pizzas based on revenue 
top = df.groupby('pizza_name')['total_price'].sum().sort_values(ascending=False).head(5)
print(top)
```

```python
x = top.index
y = top.values
plt.barh(x,y,color="purple")
plt.title("Top 5 pizzas based on revenue")
plt.xlabel("Total revenue")
plt.ylabel("Pizza name")
plt.show()
```

```python
# find orders based on quantity
n = df.groupby('quantity')['order_id'].count()
print(n)
```

```python
n = df.groupby('quantity')['order_id'].count()
x = n.index
y = n.values
plt.pie(y,labels=x,autopct='%1.2f%%')
plt.title("Orders based on quantity")
plt.show()
```

```python
# Revenue generated by each pizza
x = df.groupby('pizza_name')['total_price'].sum()
print(x)
```

```python
# Top 10 pizzas by sales quantity
top = df.groupby('pizza_name')['quantity'].sum().sort_values(ascending=False).head(10)
print(top)
```

```python
# bottom 10 pizzas by sales quantity
bottom = df.groupby('pizza_name')['quantity'].sum().sort_values(ascending=False).tail(10)
print(bottom)
```

```python
# top 5 pizzas based on revenue 
top = df.groupby('pizza_name')['total_price'].sum().sort_values(ascending=False).head(5)
print(top)
```

```python
# bottom 5 pizzas based on revenue 
bottom = df.groupby('pizza_name')['total_price'].sum().sort_values(ascending=False).tail(5)
print(bottom)
```

```python
# find highest revenue generating category
x = df.groupby('pizza_category')['total_price'].max().head(1)
print(x)
```

```python
# find highest revenue generating size
x = df.groupby('pizza_size')['total_price'].max().head(1)
print(x)
```

```python
# find highest ordered size
x = df.groupby('pizza_size')['order_id'].max()
print(x)
```

```python
# total revenue by day of week
r = df.groupby('day_name')['total_price'].sum()
print(r)
```

```python
x = r.index
y = r.values
plt.plot(x,y,'rD--')
plt.title("Total revenue by day of week")
plt.xlabel("Week Days")
plt.ylabel("Total revenue")
plt.show()
```
